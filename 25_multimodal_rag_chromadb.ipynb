# Extract Text + Images

```python
import fitz
import os

pdf_path = "/content/LSE_HWDN_2024_removed (1).pdf"
image_dir = "images"

os.makedirs(image_dir, exist_ok=True)

doc = fitz.open(pdf_path)

texts = []
images = []

for page_number, page in enumerate(doc):

    # Extract text
    text = page.get_text()

    if text.strip():
        texts.append({
            "text": text,
            "page": page_number + 1
        })

    # Extract images
    for image_number, image in enumerate(page.get_images(full=True)):

        xref = image[0]
        image_data = doc.extract_image(xref)

        image_bytes = image_data["image"]
        extension = image_data["ext"]

        image_path = (
            f"{image_dir}/"
            f"page_{page_number+1}_image_{image_number+1}.{extension}"
        )

        with open(image_path, "wb") as f:
            f.write(image_bytes)

        images.append({
            "path": image_path,
            "page": page_number + 1
        })
```

## Chunk Text

```python
!pip install langchain-text-splitters
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

splitter = RecursiveCharacterTextSplitter(
    chunk_size=800,
    chunk_overlap=100
)

chunks = []

for item in texts:

    split_chunks = splitter.split_text(item["text"])

    for chunk in split_chunks:

        chunks.append({
            "text": chunk,
            "page": item["page"],
            "type": "text"
        })
```

## Image → Vision LLM → Description

```python
import cohere
import base64

co = cohere.ClientV2(COHERE_API_KEY)
```

```python
def encode_image(image_path):

    with open(image_path, "rb") as f:
        image_bytes = f.read()

    return base64.b64encode(image_bytes).decode("utf-8")
```

```python
def describe_image(image_path):

    base64_image = encode_image(image_path)

    response = co.chat(
        model="command-a-vision-07-2025",
        messages=[
            {
                "role": "user",
                "content": [
                    {
                        "type": "text",
                        "text": """
Analyze this image for a RAG system.

If it is a chart:
- identify the title
- identify axes
- extract important values
- describe trends and comparisons

If it is a table:
- extract the important rows and columns
- preserve relationships between values

If it is a diagram:
- explain the components and relationships.

Return a concise but detailed description that can be used for semantic search.
"""
                    },
                    {
                        "type": "image_url",
                        "image_url": {
                            "url": f"data:image/png;base64,{base64_image}"
                        }
                    }
                ]
            }
        ]
    )

    return response.message.content[0].text
```

```python
image_documents = []

for image in images:

    description = describe_image(image["path"])

    image_documents.append({
        "text": description,
        "page": image["page"],
        "type": "image",
        "image_path": image["path"]
    })
```

```python
documents = chunks + image_documents
```

```python
documents
```

## Embeddings

```python
response = co.embed(
    model="embed-v4.0",
    texts=[doc["text"] for doc in documents],
    input_type="search_document",
    embedding_types=["float"]
)

embeddings = response.embeddings.float
```

## Chroma

```python
import chromadb

client = chromadb.PersistentClient(
    path="./chroma_db"
)

collection = client.get_or_create_collection(
    name="multimodal_rag"
)
```

```python
collection.add(
    ids=[str(i) for i in range(len(documents))],

    documents=[
        doc["text"]
        for doc in documents
    ],

    embeddings=embeddings,

    metadatas=[
        {
            "page": doc["page"],
            "type": doc["type"],
            "image_path": doc.get("image_path", "")
        }
        for doc in documents
    ]
)
```

```python
query = " what are Our culture and values"

response = co.embed(
    model="embed-v4.0",
    texts=[query],
    input_type="search_query",
    embedding_types=["float"]
)

query_embedding = response.embeddings.float[0]
```

## Search Chroma

```python
results = collection.query(
    query_embeddings=[query_embedding],
    n_results=5
)
```

```python
retrieved_text = results["documents"][0]
```

```python
retrieved_text
```

```python
image_path = results["metadatas"][0][0]["image_path"]
```

```python
messages_content = [
    {
        "type": "text",
        "text": f"""
Answer the question using ONLY the retrieved evidence.

Question:
{query}

Retrieved text:
{retrieved_text}
"""
    }
]

if image_path:
    messages_content.append(
        {
            "type": "image_url",
            "image_url": {
                "url": f"data:image/png;base64,{encode_image(image_path)}"
            }
        }
    )

response = co.chat(
    model="command-a-vision-07-2025",
    messages=[
        {
            "role": "user",
            "content": messages_content
        }
    ]
)
```
