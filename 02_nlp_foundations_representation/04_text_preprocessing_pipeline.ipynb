Task 1: Tokenization (Word & Sentence Level)
Goal: Understand how to split raw text into clean tokens at both sentence and word levels.

```python
import nltk
nltk.download('punkt_tab')

from nltk.tokenize import word_tokenize, sent_tokenize

text = "Hello students! Welcome to NLP Session 2. We're learning preprocessing today, aren't we?"

# 1. Split into sentences
sentences = sent_tokenize(text)
print("sentences: ",sentences)
# 2. Split into words
words = nltk.word_tokenize(text)
print("words: ",words)
```

Task 2: Regular Expressions (re Cleaning)
Goal: Clean noise (URLs, special characters, numbers, and extra spaces) from messy text before passing it to NLP models.

```python
import re

raw_tweet = "Great session today!! Check out http://example.com for NLP code @student123 #AI2026"

words = re.findall(r'\b[a-zA-Z\u0600-\u06FF]+\b', raw_tweet)
print("words: ", words)
cleaned = " ".join(words)

print("raw_tweet after cleaned: ",cleaned)
```

Task 3: Stop Words Removal
Goal: Filter out frequent, low-information words ("the", "is", "in") to focus on content-bearing tokens.

```python
import nltk
nltk.download('stopwords')
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

stop_words = set(stopwords.words('english'))

text = "This is a simple demo showing how stop words are filtered out from a sentence."

words = word_tokenize(text)
print(len(words))
print("words", words)

filtered_words = [word for word in words if not word in stop_words]
print(len(filtered_words))

print("filtered_words: ",filtered_words)
```

Task 4: Stemming vs. Lemmatization
Goal: See the difference between crude suffix stripping (Stemming) and dictionary/grammar-based root extraction (Lemmatization).

```python
import nltk
nltk.download('wordnet')
nltk.download('omw-1.4')

from nltk.stem import PorterStemmer, WordNetLemmatizer

words = ["walking", "jumps", "jumped", "jumping" ,"easily","running", "cats", "jumping", "happily","caring"]
print("words", words)

stemmer = PorterStemmer()
stemmed=[]
for word in words:
    stemmed_word = stemmer.stem(word)
    stemmed.append(stemmed_word)

print("stemms: ", stemmed)

lemmatizer = WordNetLemmatizer()

lemmas = [lemmatizer.lemmatize(word) for word in words]

print("lemmas:", lemmas)
```

```python
from nltk.stem import PorterStemmer, WordNetLemmatizer
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

text = """Natural Language Processing algorithms cannot process raw, unstructured text
 directly, making text preprocessing a fundamental prerequisite in the AI pipeline.
  Raw textual data is inherently noisy, containing irregular formatting, punctuation,
  numbers, and irrelevant web artifacts, which must first be cleaned using regular
  expressions (re). Once sanitized, tokenization breaks the text stream down into
  manageable units—either individual sentences or standalone words. To reduce
  dimensionality and focus on core semantic meaning, stop word removal filters out
  frequent yet low-information words like prepositions and conjunctions. Next,
  normalization techniques such as stemming and lemmatization reduce inflected word
  forms to their common base, with lemmatization leveraging morphological analysis to
  preserve grammatical accuracy. Finally, Part-of-Speech (POS) tagging assigns structural
   roles—such as nouns, verbs, and adjectives—to each token, resolving linguistic
   ambiguities and transforming chaotic text into clean, structured feature vectors ready
    for downstream machine learning tasks."""

words = word_tokenize(text)
print("ًWords without Regular Expressions: ", words)

words = re.findall(r'\b[a-zA-Z\u0600-\u06FF]+\b', text)
print("Words after Regular Expressions:   ", words)

stop_words = set(stopwords.words('english'))
filtered_words = [word for word in words if not word in stop_words]
print("filtered_words:                    ", filtered_words)

stemmer = PorterStemmer()
stemmed=[]
for word in words:
    stemmed_word = stemmer.stem(word)
    stemmed.append(stemmed_word)

print("stemms:                            ", stemmed)

lemmatizer = WordNetLemmatizer()
lemmas = [lemmatizer.lemmatize(word) for word in filtered_words]
print("lemmas:                            ", lemmas)
```
