#  ex3 Building a Vocabulary & Tokenization

```python
text = "ai is great and nlp is great"

# Step 1: Split text into words (Tokenization)
tokens = text.lower().split()
print("Tokens:", tokens)

# Step 2: Extract unique words to create a Vocabulary
vocab = list(set(tokens)) #vocab= list(set(tokens)) # {ai is great and nlp} ==> [ai is great  and nlp ]
vocab.sort()  # [ai and  great is nlp]
print("Vocabulary:", vocab)

# Step 3: Create Word-to-Index Mapping
word_to_id = {word: index for index, word in enumerate(vocab)}
print("Word to ID Mapping:", word_to_id) # [ai:0  aand:1  great:2 .....]

# Step 4: Convert original text to numbers
numerical_text = [word_to_id[token] for token in tokens]
print("Numerical representation:", numerical_text)
```
