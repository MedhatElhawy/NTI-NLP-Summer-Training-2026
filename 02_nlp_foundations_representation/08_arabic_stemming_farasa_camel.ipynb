# **Stemming & Lemmatization with NLTK**

```python
!pip install -qU nltk
```

```python
import nltk
from nltk.stem import PorterStemmer, SnowballStemmer

nltk.download('punkt')

porter = PorterStemmer()
snowball = SnowballStemmer("english")

words = ["running", "runner", "runs", "easily", "fairly", "happily","caring"]

print("Word\t\tPorter\t\tSnowball")
print("-" * 40)
for word in words:
    print(f"{word}\t\t{porter.stem(word)}\t\t{snowball.stem(word)}")
```

```python
import nltk
from nltk.stem import WordNetLemmatizer

nltk.download('wordnet')

lemmatizer = WordNetLemmatizer()

# Sample words
words = ["easily","running", "cats", "jumping", "happily","caring"]

lemmas = [lemmatizer.lemmatize(word, pos='v') for word in words]
lemmas_noun = [lemmatizer.lemmatize(word, pos='n') for word in words]

print("Verb lemmas:", lemmas)
print("Noun lemmas:", lemmas_noun)
```

# **Hard stemming with NLTK's ISRIStemmer**

```python
from nltk.stem.isri import ISRIStemmer

stemmer = ISRIStemmer()

arabic_words = ["يكتب", "كتب", "كُتِبَ", "مكتوب", "كاتِب", "مُكتَبات"]

print("Original\tStem")
print("-" * 30)
for word in arabic_words:
    print(f"{word}\t\t{stemmer.stem(word)}")
```

# **Light stemming with Tashaphyne**

```python
!pip install -qU tashaphyne
```

```python
from tashaphyne.stemming import ArabicLightStemmer

ar_light_stemmer = ArabicLightStemmer()

arabic_words = ["الكاتب", "مكتبة", "الكتاب", "مكتوبون"]

stemmed_words = [ar_light_stemmer.light_stem(word) for word in arabic_words]

print(f"Original words: {arabic_words}")
print(f"Stemmed words (Tashaphyne): {stemmed_words}")
```

```python
from farasa.stemmer import FarasaStemmer

stemmer = FarasaStemmer()

text = "'احلا واسرع واجود تطبيق للخدمه والاسعار واهم شي سريع خصوصا وانت جوعان مع طلبات طلباتك توصل زى ماطلبت انا احب طلباتى شكرا لخدماتكم لنا طلبات'"

stemmed_text = stemmer.stem(text)

print("Original Text:", text)
print("Stemmed Text:", stemmed_text)
```

```python
from camel_tools.morphology.database import MorphologyDB
from camel_tools.morphology.analyzer import Analyzer

db = MorphologyDB.builtin_db('calima-msa-r13')
analyzer = Analyzer(db)

tokens = 'احلا واسرع واجود تطبيق للخدمه والاسعار واهم شي سريع خصوصا وانت جوعان مع طلبات طلباتك توصل زى ماطلبت انا احب طلباتى شكرا لخدماتكم لنا طلبات'.split()
for token in tokens:
    analyses = analyzer.analyze(token)
    if analyses:
        print(f"  Lemma: {analyses[0]['lex']} — stem: {analyses[0]['stem']}")
    else:
        print(f"  No analysis for '{token}'")
```
