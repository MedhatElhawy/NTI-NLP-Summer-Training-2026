Tokenizer

```python
from tensorflow.keras.preprocessing.text import Tokenizer

texts = [
    "I love NLP",
    "I love Machine Learning",
    "Deep Learning is amazing"
]

tokenizer = Tokenizer()

tokenizer.fit_on_texts(texts)

print(tokenizer.word_index)
```

# Convert Text to Sequences

```python
sequences = tokenizer.texts_to_sequences(texts)

print(sequences)
```

# Padding

```python
from tensorflow.keras.preprocessing.sequence import pad_sequences

padded = pad_sequences(
    sequences,
    maxlen=6,
    padding="post"
)

print(padded)
```

# Simple Data set

```python
texts = [
    "good movie",
    "excellent film",
    "awesome movie",
    "bad movie",
    "terrible film",
    "worst movie"
]

labels = [1,1,1,0,0,0]
```

Tokenizer

```python
tokenizer = Tokenizer()

tokenizer.fit_on_texts(texts)

sequences = tokenizer.texts_to_sequences(texts)

X = pad_sequences(sequences,maxlen=2)

print(X)
```

# Build RNN

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding,SimpleRNN,Dense

model = Sequential()

model.add(
    Embedding(
        input_dim=20,
        output_dim=8,
        input_length=2
    )
)

model.add(SimpleRNN(16))

model.add(Dense(1,activation="sigmoid"))
```

```python
model.summary()
```

```python
model.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy"]
)
```

```python
import numpy as np

model.fit(
    X,
    np.array(labels),
    epochs=30,
    verbose=1
)
```

Prediction

```python
test = tokenizer.texts_to_sequences(
    ["good film"]
)

test = pad_sequences(test,maxlen=2)

prediction = model.predict(test)

print(prediction)
```

```python
test = tokenizer.texts_to_sequences(
    ["worst film"]
)

test = pad_sequences(test,maxlen=2)

prediction = model.predict(test)

print(prediction)
```
