# Part 1 - Word2Vec

```python
import nltk


texts = [
    "AI is changing the world",
    "Machine learning is amazing",
    "Deep learning powers many applications",
    "I enjoy learning NLP",
    "NLP is fun and useful"
]

tokens = [nltk.word_tokenize(text) for text in texts]
print(tokens)
```

## Task1
 Train a Word2Vec model using vector_size=30, window=2, min_count=1.

```python
from gensim.models import Word2Vec

model = Word2Vec(tokens, vector_size=50, window=3, min_count=1)
```

## Task2
 Find the number of unique words and identify the most frequent word without printing the entire vocabulary.

```python
number_of_unique_words = len(model.wv)
print("Number of unique words: ", number_of_unique_words)

most_frequent_word = model.wv.index_to_key[0]
print("Most frequent word: ", most_frequent_word)
```

## Task3
 Print the word vectors of any two words of your choice.

```python
# word1 'NLP'
print('word vector for word "NLP": ',model.wv['NLP'])

# word2 'and'
print('word vector for word "and":',model.wv['and'])
```

## Task4
 Find the two most similar words to "learning".

```python
similar_words = model.wv.similar_by_word('learning', topn=2)

print(similar_words)
```

## Task5
 Choose any two pairs of words, calculate their similarity, and briefly explain the results.

 explain the results:
  Cosine Similarity Scale: Values range from $-1.0$  to $1.0$.

  Machine vs. learning ($sim_1$) High/Positive Similarity  -> 0.15019882

  enjoy vs. NLP ($sim_2$)Lower Similarity -> 0.08872843

```python
pair1 = ['Machine', 'learning']
pair2 = ['enjoy', 'NLP']

sim1 = model.wv.similarity(pair1[0],pair1[1])
sim2 = model.wv.similarity(pair2[0],pair2[1])

print('Similarity between "Machine", "learning": ', sim1)
print('Similarity between "enjoy", "NLP": ', sim2)
```

# Part 2 - Simple RNN

```python
texts = [
    "I love AI",
    "AI is amazing",
    "I hate bugs",
    "Coding is fun",
    "This movie is terrible",
    "This movie is great"
]

labels = [1, 1, 0, 1, 0, 1]
```

## Task1
 Convert the text into sequences using TensorFlow Tokenizer and print only the sequences.

```python
from tensorflow.keras.preprocessing.text import Tokenizer
import numpy as np
import tensorflow as tf
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, SimpleRNN, Dense

tokenizer = Tokenizer()
tokenizer.fit_on_texts(texts)
print('word_index: ',tokenizer.word_index)

sequences = tokenizer.texts_to_sequences(texts)
print('texts_to_sequences: ',sequences)
```

## Task2
  Pad all sequences to the maximum sentence length in the dataset.

```python
pad = pad_sequences(
    sequences,
    maxlen=5,
    padding='post'
    
)
print(pad)
```

## Task3
 Build a model with Embedding, SimpleRNN, and Dense output layers. Choose the embedding dimension and the number of RNN units yourself.

```python
vocab_size = len(tokenizer.word_index) + 1

model = Sequential([
    Embedding(input_dim=vocab_size, output_dim=8, input_length=5),
    SimpleRNN(units=16),
    Dense(1, activation="sigmoid")
])
```

## Task4
Train the model for 15 epochs.

```python
model.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy"]
)

history = model.fit(
    pad,
    np.array(labels),
    epochs=15,
    verbose=1
)
```

## Task5
 Predict the sentiment of any two new sentences of your choice and explain whether the predictions are reasonable.

```python
test_sentences = [
    "movie is amazing", 
    "hate bugs"
]

test_seq = tokenizer.texts_to_sequences(test_sentences)
test_padded = pad_sequences(test_seq, maxlen=5, padding='post')

predictions = model.predict(test_padded)
print(predictions)
```
