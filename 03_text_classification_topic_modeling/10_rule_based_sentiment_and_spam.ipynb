Exercise: "Build a Toy Sentiment Classifier"

Task: Students use positive and negative word lists to score a review.

---



```python
# Define sentiment vocabularies
positive_words = ["love", "great", "awesome", "fast", "good"]
negative_words = ["bad", "slow", "terrible", "hate", "worst"]

# Input sentence
review = "The camera is great and awesome but the battery is bad"


def classify_sentiment(review):
    score = 0
    words = review.lower().split()

    for word in words:
        if word in positive_words:
            score += 1
        elif word in negative_words:
            score -= 1
    return score

sentiment_score = classify_sentiment(review)
if sentiment_score > 0:
    sentiment = "Positive"
elif sentiment_score < 0:
    sentiment = "Negative"
else:
    sentiment = "Neutral"

print(f"Sentiment: {sentiment}")
```

Example 2: Application (Spam Filter / Text Classification)
Concept: A basic rule-based application demonstrating how NLP systems classify text using predefined vocabulary checks.

```python
# Define vocabulary rules for spam detection
spam_keywords = ["free", "win", "cash", "click", "prize", "urgent"]

threshold = 2
def classify_email(email_text):

    words = email_text.lower().split()

    score = sum(1 for word in words if word in spam_keywords)
    return "Spam" if score >= threshold else "Not Spam"

email_text = """
Subject: Urgent: Account Suspended! Your account has been temporarily restricted due to suspicious activity.
Click here to verify your cash balance and win back access immediately.
"""

classify_email(email_text)
```
